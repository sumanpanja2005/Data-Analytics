```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\suman\OneDrive\download\OneDrive\Desktop\Data Science Project\global_superstore_2016(Orders).csv",
    encoding="latin1"
)
```

```python
num_dupes = df.duplicated().sum()
print(f"Total duplicates: {num_dupes}")
df = df.drop_duplicates()
```

```python
print(df.isna().sum())
```

```python
print((df.isna().mean() * 100).round(2))
df.to_csv("dataset_cleaned.csv", index=False)
```

```python
print(df.isna().sum())
```

```python
pivot = df.pivot_table(
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0,      # replaces NaNs with 0
    margins=True,       # adds 'All' grand totals row/col
    margins_name="Total"
)
```

```python
pivot.reset_index().to_csv("region_wise_sales.csv", index=False)
```
