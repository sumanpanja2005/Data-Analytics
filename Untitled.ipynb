```python
pivot = df.pivot_table(
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0,      # replaces NaNs with 0
    margins=True,       # adds 'All' grand totals row/col
    margins_name="Total"
)
```

```python
pivot.reset_index().to_csv("region_wise_sales.csv", index=False)
```
